```python
import pandas as pd
```

```python
customers = pd.read_csv("../Data/raw/olist_customers_dataset.csv")

orders = pd.read_csv(
    "../Data/raw/olist_orders_dataset - olist_orders_dataset.csv"
)

order_items = pd.read_csv("../Data/raw/olist_order_items_dataset.csv")

payments = pd.read_csv(
    "../Data/raw/olist_order_payments_dataset - olist_order_payments_dataset.csv"
)

reviews = pd.read_csv(
    "../Data/raw/olist_order_reviews_dataset - olist_order_reviews_dataset.csv"
)

products = pd.read_csv(
    "../Data/raw/olist_products_dataset - olist_products_dataset.csv"
)

sellers = pd.read_csv(
    "../Data/raw/olist_sellers_dataset - olist_sellers_dataset.csv"
)

geolocation = pd.read_csv(
    "../Data/raw/olist_geolocation_dataset - olist_geolocation_dataset.csv"
)

category_translation = pd.read_csv(
    "../Data/raw/product_category_name_translation - product_category_name_translation.csv"
)

datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Products": products,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}
```

```python
for name, df in datasets.items():
    print("=" * 70)
    print(name)

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(missing)
```

```python
for name, df in datasets.items():
    print(
        name,
        "→ Duplicate rows:",
        df.duplicated().sum()
    )
```

```python
numeric_checks = {
    "Order Items - price": order_items["price"],
    "Order Items - freight_value": order_items["freight_value"],
    "Payments - payment_value": payments["payment_value"],
    "Payments - payment_installments": payments["payment_installments"],
    "Products - product_weight_g": products["product_weight_g"],
    "Products - product_length_cm": products["product_length_cm"],
    "Products - product_height_cm": products["product_height_cm"],
    "Products - product_width_cm": products["product_width_cm"]
}

for name, series in numeric_checks.items():
    print(
        name,
        "→ Negative values:",
        (series < 0).sum()
    )
```

```python
print(reviews["review_score"].value_counts(dropna=False).sort_index())
invalid_reviews = (
    (reviews["review_score"] < 1) |
    (reviews["review_score"] > 5)
).sum()

print("Invalid review scores:", invalid_reviews)
```

```python
quality_report = []

for name, df in datasets.items():
    quality_report.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

quality_df = pd.DataFrame(quality_report)

display(quality_df)
```
